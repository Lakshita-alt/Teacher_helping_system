```python
core_features = [
 "Hours_Studied",
 "Attendance",
 "Sleep_Hours",
 "Motivation_Level",
 "Tutoring_Sessions",
 "Physical_Activity",
 "Access_to_Resources",
 "Internet_Access",
 "Peer_Influence",
 "Extracurricular_Activities"
]
```

```python
for col in df.columns:
    if df[col].isna().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)

print(df.isna().sum())
```

```python
cat_col = ["Motivation_Level",
 "Access_to_Resources",
 "Internet_Access",
 "Peer_Influence",
 "Extracurricular_Activities",
 "Parental_Involvement",
 "Family_Income",
 "Teacher_Quality",
 "School_Type",
 "Learning_Disabilities",
 "Parental_Education_Level",
 "Distance_from_Home",
 "Gender"]
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df_encoded_le = df.copy()
for col in cat_col:
    df_encoded_le[col] = le.fit_transform(df_encoded_le[col])
```

```python
df_encoded_le.head()
```

```python
import numpy as np 

binary_features = ["Extracurricular_Activity", 
                   "Internet_Access",
                   "School_Type",
                   "Learning_Disabilities",
                   "Gender"]

for col in core_features:
    if col not in binary_features:
        q1 = df_encoded_le[col].quantile(0.25)
        q3 = df_encoded_le[col].quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        df_encoded_le[col] = np.where(df_encoded_le[col] < lower, lower, df_encoded_le[col])
        df_encoded_le[col] = np.where(df_encoded_le[col] > upper, upper, df_encoded_le[col])

df_encoded_le.describe ()
```

```python
for col in df_encoded_le.columns:
    print(col)
    print(df_encoded_le[col].var())
    print("=" * 10)
```

```python
# From the previous experiment we can derive that Internet_Access has extremely low variance and is not really that helpful for the clustering
# So we will remove it from the core features.
core_features.remove("Internet_Access")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 10))

corr = df_encoded_le[core_features].corr()
sns.heatmap(corr, annot=True)
plt.show()
```

```python
# Heatmap clearly shows that the features are not much correlated

from sklearn.preprocessing import StandardScaler

X = df_encoded_le[core_features]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []
for k in range(2, 8):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(2,8), wcss)
plt.xlabel("Number of clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

```python
# From the elbow graph, we can see that the optimal number of clusters is 4.

from sklearn.metrics import silhouette_score

for k in range(2,8):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    print(k, silhouette_score(X_scaled, labels))
```

```python
# According to the silhouette score, Kmeans might not be fit for this data.

kmeans = KMeans(n_clusters=4, random_state=42)
df_encoded_le["KMeans_Cluster"] = kmeans.fit_predict(X_scaled)
```

```python
# After trying different clustering algorithms, we decided to use Hiearchical clustering

persona_map = {
    1: "Average but Low-Engagement Learner",
    2: "Motivated but Under-Resourced Learner",
    3: "Well-Supported Consistent Learner"
}

df_encoded_le["Persona"] = df_encoded_le["H_Cluster"].map(persona_map)
```

```python
# Since personas alone were not enough to predict the outcome, we used risk scores to help with the oucome 

df_encoded_le["Risk_Score"] = 1 - (
    (df_encoded_le["Attendance"] +
     df_encoded_le["Hours_Studied"] +
     df_encoded_le["Motivation_Level"]) /
    (df_encoded_le[["Attendance","Hours_Studied","Motivation_Level"]].max().sum())
)
```

```python
def risk_category(score):
    if score < 0.33:
        return "Low Risk"
    elif score < 0.66:
        return "Medium Risk"
    else:
        return "High Risk"

df_encoded_le["Risk_Category"] = df_encoded_le["Risk_Score"].apply(risk_category)
```

```python
# Function to compare the random student profile to the class averages

import random
import pandas as pd

def show_random_student_profile(df, class_means):
    # Pick a random student
    idx = random.choice(df.index)
    student = df.loc[idx]

    print("\n" + "="*60)
    print(f"EXAMPLE STUDENT PROFILE (Student #{idx})")
    print("="*60)

    print(f"\nPersona: {student['Persona']}")
    print(f"Risk Level: {student['Risk_Category']}")
    print(f"Risk Score: {round(student['Risk_Score'], 2)}")

    print("\n--- KEY METRICS (Student vs Class Average) ---\n")

    comparison_table = []

    features = [
        "Attendance",
        "Hours_Studied",
        "Sleep_Hours",
        "Motivation_Level",
        "Tutoring_Sessions",
        "Access_to_Resources",
        "Peer_Influence",
        "Extracurricular_Activities"
    ]

    for f in features:
        student_val = round(student[f], 2)
        avg_val = round(class_means[f], 2)
        diff = round(student_val - avg_val, 2)

        comparison_table.append({
            "Metric": f,
            "Student": student_val,
            "Class_Average": avg_val,
            "Difference": diff
        })

    comparison_df = pd.DataFrame(comparison_table)
    print(comparison_df.to_string(index=False))

    print("\n--- QUICK INTERPRETATION (for non-ML teammates) ---")
    print("• Positive difference = student is ABOVE class average")
    print("• Negative difference = student is BELOW class average")
    print("="*60 + "\n")

# Run it
show_random_student_profile(df_encoded_le, class_means)
```

```python
# Visualize the student vs class average

import matplotlib.pyplot as plt

def plot_student_vs_average(student, class_means):
    features = [
        "Attendance",
        "Hours_Studied",
        "Sleep_Hours",
        "Motivation_Level",
        "Tutoring_Sessions",
        "Access_to_Resources",
        "Peer_Influence",
        "Extracurricular_Activities"
    ]

    student_vals = [student[f] for f in features]
    avg_vals = [class_means[f] for f in features]

    x = range(len(features))

    plt.figure(figsize=(10,5))
    plt.bar(x, student_vals, alpha=0.6, label="Student")
    plt.bar(x, avg_vals, alpha=0.4, label="Class Average")
    plt.xticks(x, features, rotation=45)
    plt.title("Student vs Class Average")
    plt.legend()
    plt.tight_layout()
    plt.show()

# Example usage:
idx = random.choice(df_encoded_le.index)
plot_student_vs_average(df_encoded_le.loc[idx], class_means)
```
